```python
def binary_search(list, item):
    low = 0
    high = len(list)-1
    
    while low <= high:
        mid = (low + high)//2
        guess = list[mid]
        if guess == item:
            return mid
        if guess > item:
            high = mid - 1
        else:
            low = mid + 1
    return None

my_list = [1, 3, 5, 7, 9]
   
print(binary_search(my_list, 3)) # => 1
print(binary_search(my_list, -1)) # => None             

my_list.extend([11, 13]) 
   
print(binary_search(my_list, 11)) # => 1
            
```

3(a) Naming the function parameter list shadows Python’s built-in list data type, which is generally discouraged in best practices. Additionally, mid points to the index of the middle element ((low + high) // 2), and incorrect handling of bounds can lead to infinite loops if not properly adjusted (mid - 1 and mid + 1)
3(b) Binary search relies on the sorted order. It cuts the search space in half at each step by comparing the target item to the middle element; if the list were unsorted, knowing whether the target is greater than or less than the middle element would give no logical clue as to whether it lies in the left or right half.

```python
# Define a class named Node to represent a single element in the linked list
class Node:
    # Constructor method to initialize a new node with a number value
    def __init__(self, data):
        self.data = data # Store the number value passed into the node
        self.next = None # Set the pointer to the next node as None by default

# Define a class named SinglyLinkedList to manage the collection of nodes
class LinkedList: 
    # Constructor method to initialize an empty linked list with a head pointer
    def __init__(self):
        self.head = None # Initialize the head of the linked list as None, indicating that the list is empty

    # Method to append a new node with the given data to the end of the linked list
    def append(self, data):
        new_node = Node(data) # Create a new node with the provided data

        if self.head is None: # Check if the linked list is empty (i.e., head is None)
            self.head = new_node # If the list is empty, set the head of the linked list to the new node
            return # If the list is not empty, traverse the linked list to find the last node

        current = self.head # Start traversing the linked list from the head node
        while current.next is not None: # Continue traversing the linked list until reaching the last node (i.e., a node whose next pointer is None)
            current = current.next # Move to the next node in the linked list

        current.next = new_node # After reaching the last node, set its next pointer to the new node, effectively appending the new node to the end of the linked list


 # Create an instance of the SinglyLinkedList
my_linked_list = LinkedList() 

# Append some data to the linked list
my_linked_list.append(1)
my_linked_list.append(2)
my_linked_list.append(3)



print("Linked list code executed successfully!")        
```
